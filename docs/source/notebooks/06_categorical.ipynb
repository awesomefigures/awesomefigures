# Categorical: bars, dots, and lollipops

Compact comparison figures for categorical outcomes.

```python
import numpy as np
import matplotlib.pyplot as plt
import awesomefigures as af

af.set_style("heri", font="DejaVu Sans")
labels=["A","B","C","D","E"]
values=np.array([1.2,1.7,0.95,1.45,1.1])

fig,ax=plt.subplots(figsize=(7.8,4.0))
ax.bar(labels,values,color=af.HERI_PALETTE,edgecolor="white",linewidth=0.7)
ax.set(title="Vertical bars",xlabel="Category",ylabel="Value")
fig.tight_layout()
```

```python
af.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(7.8,4.0))
ax.barh(labels,values,color=af.HERI_PALETTE)
ax.set(title="Horizontal bars",xlabel="Value",ylabel="Category")
fig.tight_layout()
```

```python
af.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(7.8,4.0))
for i,(label,value,color) in enumerate(zip(labels,values,af.HERI_PALETTE)):
    ax.plot([0,value],[i,i],color=af.HERI_NEUTRAL["light"],linewidth=1.2)
    ax.scatter(value,i,color=color,s=70,zorder=3)
ax.set_yticks(range(len(labels)),labels)
ax.set(title="Dot plot",xlabel="Value",ylabel="")
fig.tight_layout()
```

```python
af.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(7.8,4.0))
for i,(value,color) in enumerate(zip(values,af.HERI_PALETTE)):
    ax.vlines(i,0,value,color=color,linewidth=2)
    ax.scatter(i,value,color=color,s=72,zorder=3)
ax.set_xticks(range(len(labels)),labels)
ax.set(title="Lollipop plot",xlabel="Category",ylabel="Value")
fig.tight_layout()
```

```python
af.show_accessibility_panel(fig, max_width=640)
plt.show()
```
